# Pandas Concatenation
# ---------------------------------------------------



```python
# Create dataframe d1
d1 = pd.DataFrame({
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
})

# Print d1
print(d1)
```

```python
# What is the shape of d1
print(d1.shape)
```

```python

# Set city = d1['city']
city = d1['city']
```

```python
# print city
print(city)
```

```python
# What is the type of city
print(type(city))
```

```python
# Create another dataframe d2
d2 = pd.DataFrame({
    'city': ['Bengalaru', 'Coimbatore', 'Srirangam', 'Pondicherry'],
    'Temperature': [24, 35, 36, 39]
})

print(d2)
```

```python

# print the shape of this dataframe
print(d2.shape)
```

```python
# merge the two dataframes together
d3 = pd.concat([d1, d2], ignore_index=True)

print(d3)
```

```python

# Select the part of dataframe where temp <= 20
temp_less_20 = d3[d3['Temperature'] <= 20]

print(temp_less_20)
```

```python
# How many cities are there?
print(len(temp_less_20))
```

```python
temp_greater_35 = d3[d3['Temperature'] >= 35]

print(temp_greater_35)
```

# ---------------------------------------------------
# Applying functions to columns and creating new columns
# ---------------------------------------------------

```python

# Function to check Union Territory
def is_ut(x):
    ut_cities = ['Chandigarh', 'Delhi', 'Pondicherry']
    return x in ut_cities


d3['is_ut'] = d3['city'].apply(is_ut)

print(d3)
```

```python
# Function to convert Celsius to Fahrenheit
def c_to_f(temp):
    return (temp * 9/5) + 32
```

```python

# check function
print(c_to_f(10))
```

```python
# apply function to create new column
d3['temp_fahrenheit'] = d3['Temperature'].apply(c_to_f)

print(d3)
```

```python
# ---------------------------------------------------
# Indexing and selecting rows in DataFrame
# ---------------------------------------------------
```

```python
# Cities which are union territories
print(d3[d3['is_ut'] == True])
```

```python
# Cities having temperature above 90 Fahrenheit
print(d3[d3['temp_fahrenheit'] > 90])
```

```python

# Select first three rows of dataframe d1
print(d1.iloc[:3])
```

```python
# Select all rows and last two columns
print(d3.iloc[:, -2:])
```

# ---------------------------------------------------
# Groupby
# ---------------------------------------------------

```python
# Create dataframe using dictionary
data = pd.DataFrame({
    'Department': ['HR', 'HR', 'IT', 'IT', 'Finance', 'Finance'],
    'Employee': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Salary': [25000, 30000, 40000, 45000, 35000, 30000],
    'Experience': [2, 4, 5, 6, 3, 2]
})

print(data)
```

```python
# Groupby single column with aggregate sum()
print(data.groupby('Department')['Salary'].sum())
```

```python

# Groupby single column with aggregate count()
print(data.groupby('Department')['Employee'].count())
```

```python
# Groupby single column with aggregate min() and max()
print(data.groupby('Department')['Salary'].agg(['min', 'max']))
```

```python
# Groupby any 2 columns with aggregate mean()
print(data.groupby(['Department', 'Experience'])['Salary'].mean())
```

```python
# Groupby any 2 columns with aggregate min() and max()
print(data.groupby(['Department', 'Experience'])['Salary'].agg(['min', 'max']))
```
